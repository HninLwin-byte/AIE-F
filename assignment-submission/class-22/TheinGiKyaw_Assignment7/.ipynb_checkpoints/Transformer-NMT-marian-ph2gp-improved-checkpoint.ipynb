# Transformer Marian ph->my Improvement Experiments

This notebook creates new Transformer experiment folders and uses ideas from other Class 22 submissions:

- The standard Transformer baseline reached around 76 BLEU in several submissions.
- Aung Chan Nyein's Transformer v2 uses a shared vocabulary for `tied-embeddings-all`, `keep-best`, learning-rate warmup, lower dropout variants, and decoding from the best BLEU checkpoint.

For this WSL laptop setup, the config uses one GPU (`devices: [0]`) and disables CPU training (`cpu-threads: 0`).

```python
from pathlib import Path
import os
import re
import subprocess
import textwrap

WS = Path.cwd()
DATA = WS / "g2p-par"
DEVICE = os.environ.get("MARIAN_DEVICES", "0")

marian_candidates = [
    os.environ.get("MARIAN_BIN"),
    "/mnt/d/Job_Requirements/PracticalProjectsFromDrYe/marian/build/marian",
    "/d/Job_Requirements/PracticalProjectsFromDrYe/marian/build/marian",
    "marian",
]

MARIAN_BIN = next((p for p in marian_candidates if p and (Path(p).exists() or p == "marian")), None)
MARIAN_DECODER = str(Path(MARIAN_BIN).with_name("marian-decoder")) if MARIAN_BIN != "marian" else "marian-decoder"
MARIAN_VOCAB = str(Path(MARIAN_BIN).with_name("marian-vocab")) if MARIAN_BIN != "marian" else "marian-vocab"

print("Workspace:", WS)
print("Marian:", MARIAN_BIN)
print("Decoder:", MARIAN_DECODER)
print("Vocab:", MARIAN_VOCAB)
print("GPU device:", DEVICE)

subprocess.run([MARIAN_BIN, "--build-info"], check=True)
```

## Helpers

```python
def run(cmd, stdin=None):
    print("$", " ".join(map(str, cmd)))
    return subprocess.run(cmd, stdin=stdin, text=True, check=True)


def find_multi_bleu():
    candidates = [
        WS / "multi-bleu.perl",
        Path.home() / "tools/mosesbin/ubuntu-17.04/mosesdecoder/scripts/generic/multi-bleu.perl",
        Path("/home/gi/tools/mosesbin/ubuntu-17.04/mosesdecoder/scripts/generic/multi-bleu.perl"),
        Path("/home/gi/tools/mosesbin/ubuntu-17.04/mosesdecoder/scripts/generic/multi-bleu.perl"),
    ]
    for p in candidates:
        if p.exists():
            return str(p)
    return None


def parse_bleu(text):
    m = re.search(r"BLEU = ([0-9.]+)", text)
    return float(m.group(1)) if m else None


def best_dev_bleu(valid_log):
    valid_log = Path(valid_log)
    best, best_update = None, None
    if not valid_log.exists():
        return None, None
    for line in valid_log.read_text(encoding="utf-8", errors="ignore").splitlines():
        m = re.search(r"Up\.\s+(\d+).*?bleu\s*:\s*([0-9.]+)", line)
        if m:
            update, bleu = int(m.group(1)), float(m.group(2))
            if best is None or bleu > best:
                best, best_update = bleu, update
    return best, best_update


def evaluate_bleu(ref_file, hyp_file):
    multi_bleu = find_multi_bleu()
    if not multi_bleu:
        print("multi-bleu.perl not found. Install/download it, then rerun this cell.")
        return None
    with open(hyp_file, encoding="utf-8", errors="ignore") as hyp:
        result = subprocess.run(["perl", multi_bleu, str(ref_file)], stdin=hyp, capture_output=True, text=True, check=True)
    print(result.stdout)
    return parse_bleu(result.stdout)
```

## Build Shared Vocabulary

`vocab.shared.yml` is required for `tied-embeddings-all: true`, because source and target vocabularies must have the same size.

```python
shared_vocab = DATA / "vocab/vocab.shared.yml"
shared_vocab.parent.mkdir(parents=True, exist_ok=True)
combined_text = "\n".join([
    (DATA / "train.ph").read_text(encoding="utf-8", errors="ignore"),
    (DATA / "train.my").read_text(encoding="utf-8", errors="ignore"),
    (DATA / "dev.ph").read_text(encoding="utf-8", errors="ignore"),
    (DATA / "dev.my").read_text(encoding="utf-8", errors="ignore"),
])
result = subprocess.run([MARIAN_VOCAB], input=combined_text, capture_output=True, text=True, check=True)
shared_vocab.write_text(result.stdout, encoding="utf-8")
print("Shared vocab:", shared_vocab)
print("Entries:", sum(1 for _ in shared_vocab.open(encoding="utf-8")))
```

## Create Transformer Configs

Run variants one by one and compare validation/test BLEU.

```python
def write_transformer_config(name, dropout=0.3, warmup=0):
    model_dir = WS / f"model.transformer.{name}.phmy"
    model_dir.mkdir(parents=True, exist_ok=True)
    config = f"""
type: transformer
train-sets:
  - {DATA}/train.ph
  - {DATA}/train.my
valid-sets:
  - {DATA}/dev.ph
  - {DATA}/dev.my
vocabs:
  - {shared_vocab}
  - {shared_vocab}
model: {model_dir}/model.npz
max-length: 100
mini-batch-fit: true
workspace: 3000
maxi-batch: 100
transformer-heads: 8
transformer-dropout: {dropout}
transformer-dim-ffn: 2048
transformer-ffn-activation: swish
transformer-postprocess-emb: d
transformer-postprocess: dan
enc-depth: 2
dec-depth: 2
tied-embeddings-all: true
learn-rate: 0.0003
lr-decay-inv-sqrt: 16000
lr-warmup: {warmup}
lr-report: true
label-smoothing: 0.1
clip-norm: 5
valid-freq: 5000
save-freq: 5000
disp-freq: 500
valid-metrics:
  - cross-entropy
  - perplexity
  - bleu
early-stopping: 10
beam-size: 6
normalize: 0.6
log: {model_dir}/train.log
valid-log: {model_dir}/valid.log
valid-translation-output: {model_dir}/valid.ph-my.output
overwrite: true
keep-best: true
seed: 1111
devices:
  - {DEVICE}
cpu-threads: 0
quiet-translation: true
""".strip() + "\n"
    config_path = model_dir / "config.yml"
    config_path.write_text(config, encoding="utf-8")
    return model_dir, config_path

experiments = {
    "baseline_like": dict(dropout=0.3, warmup=0),
    "warmup4000": dict(dropout=0.3, warmup=4000),
    "dropout01": dict(dropout=0.1, warmup=0),
    "warmup4000_dropout01": dict(dropout=0.1, warmup=4000),
}

for name, params in experiments.items():
    model_dir, config_path = write_transformer_config(name, **params)
    print(name, "->", config_path)
```

## Train One Experiment

Change `EXP_NAME` to run a different variant.

```python
EXP_NAME = "warmup4000_dropout01"
model_dir, config_path = write_transformer_config(EXP_NAME, **experiments[EXP_NAME])
run([MARIAN_BIN, "--config", str(config_path)])
```

## Decode and Evaluate

```python
EXP_NAME = "warmup4000_dropout01"
model_dir = WS / f"model.transformer.{EXP_NAME}.phmy"
best_model = model_dir / "model.npz.best-bleu.npz"
if not best_model.exists():
    best_model = model_dir / "model.npz"

hyp_file = WS / f"transformer.{EXP_NAME}.phmy.hyp.txt"
cmd = f'{MARIAN_DECODER} -m {best_model} -v {shared_vocab} {shared_vocab} --devices {DEVICE} < {DATA}/test.ph > {hyp_file}'
print("$", cmd)
subprocess.run(cmd, shell=True, check=True)
print("Saved:", hyp_file)
evaluate_bleu(DATA / "test.my", hyp_file)
```

## Compare Runs

```python
rows = []
for name in experiments:
    model_dir = WS / f"model.transformer.{name}.phmy"
    hyp_file = WS / f"transformer.{name}.phmy.hyp.txt"
    dev_bleu, dev_update = best_dev_bleu(model_dir / "valid.log")
    test_bleu = evaluate_bleu(DATA / "test.my", hyp_file) if hyp_file.exists() else None
    rows.append((name, test_bleu, dev_bleu, dev_update, model_dir))

print("\nTransformer summary")
print(f'{"experiment":<26} {"test BLEU":>10} {"best dev":>10} {"update":>10}')
for name, test_bleu, dev_bleu, dev_update, model_dir in rows:
    print(f'{name:<26} {test_bleu if test_bleu is not None else "N/A":>10} {dev_bleu if dev_bleu is not None else "N/A":>10} {dev_update if dev_update is not None else "N/A":>10}')
```
